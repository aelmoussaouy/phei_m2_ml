Quick test: fill a Gaussian histogram, draw it and save it as PNG

```python
h = ROOT.TH1F("h", "Gaussian test;x;Entries", 50, -4, 4)
h.FillRandom("gaus", 10000)
print(f"Entries: {h.GetEntries():.0f}, mean: {h.GetMean():.3f}, std dev: {h.GetStdDev():.3f}")

c = ROOT.TCanvas("c", "", 600, 400)
h.Draw()
c.Draw()
c.SaveAs("gaussian_test.png")
```
